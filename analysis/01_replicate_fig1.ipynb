# Replicating Figure 1: Age vs Hubble Residual

This notebook attempts to replicate Figure 1 from the paper, specifically the correlation between host galaxy age and Hubble Residual (HR) for the Rose et al. (2019) sample.

## 1. Load Data
We load the Rose et al. (2019) data downloaded from VizieR: Table 1 (HR) and Table 7 (Ages).

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import os
import scipy.stats as stats

# Define paths
ROSE19_TABLE1 = os.path.join('..', 'data', 'external', 'Rose19', 'J_ApJ_874_32_table1.csv')
ROSE19_TABLE7 = os.path.join('..', 'data', 'external', 'Rose19', 'J_ApJ_874_32_table7.csv')

# Load data
t1 = pd.read_csv(ROSE19_TABLE1)
t7 = pd.read_csv(ROSE19_TABLE7)

print("Table 1 (HR) columns:", t1.columns.tolist())
print("Table 7 (Age) columns:", t7.columns.tolist())

# Merge on SNID
# Ensure SNID is same type (int)
t1['SNID'] = t1['SNID'].astype(int)
t7['SNID'] = t7['SNID'].astype(int)

df = pd.merge(t1, t7, on='SNID', how='inner')
print(f"Matched {len(df)} supernovae.")
df.head()
```

## 2. Plot Age vs Hubble Residual
We use `logAl` (prob. log Age Local) as the age metric.

```python
plt.figure(figsize=(8, 6))
plt.errorbar(df['logAl'], df['HR'], yerr=df['e_HR'], fmt='o', alpha=0.7, label='Data')

plt.xlabel('Log Age (Local) [Gyr?]')
plt.ylabel('Hubble Residual (mag)')
plt.title('Hubble Residual vs Host Age (Rose+19)')
plt.axhline(0, color='k', linestyle='--', alpha=0.5)
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

## 3. Statistical Analysis
Check for correlation.

```python
# Drop NaNs if any
df_clean = df.dropna(subset=['logAl', 'HR'])

pearson_r, p_val = stats.pearsonr(df_clean['logAl'], df_clean['HR'])
print(f"Pearson r: {pearson_r:.3f}")
print(f"P-value: {p_val:.3e}")

slope, intercept, r_value, p_value, std_err = stats.linregress(df_clean['logAl'], df_clean['HR'])
print(f"Slope: {slope:.3f} +/- {std_err:.3f}")
```
