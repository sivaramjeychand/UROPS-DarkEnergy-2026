# Figure 1 Replication: The Age-Bias Correlation

This notebook replicates Figure 1, demonstrating the local correlation between host galaxy age and Hubble residuals. We use a professional plotting style with binned data for clarity.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

# 1. Load Data
DATA_FILE = 'Rose19_corrected.csv'
df = pd.read_csv(DATA_FILE)

print(f"Loaded {len(df)} SNe.")
```

```python
plt.figure(figsize=(9, 7))

# Individual Points
plt.scatter(df['logAl'], df['HR'], alpha=0.3, color='gray', s=30, label='Rose+19 Individual SNe')

# Linear Fit
slope, intercept, r_value, p_value, std_err = stats.linregress(df['logAl'], df['HR'])
x_fit = np.linspace(df['logAl'].min(), df['logAl'].max(), 100)
plt.plot(x_fit, slope*x_fit + intercept, 'k-', linewidth=2, label=f'Best-fit (Slope = {slope:.3f} $\pm$ {std_err:.3f})')

# Binned Points
bins = np.linspace(df['logAl'].min(), df['logAl'].max(), 7)
c = pd.cut(df['logAl'], bins)
m = df.groupby(c, observed=False)[['logAl']].mean()
v = df.groupby(c, observed=False)[['HR']].mean()
e = df.groupby(c, observed=False)[['HR']].std() / np.sqrt(df.groupby(c, observed=False)[['HR']].count())

plt.errorbar(m['logAl'], v['HR'], yerr=e['HR'], fmt='o', color='red', markersize=10, 
             markeredgecolor='black', label='Binned Averages', zorder=10)

plt.axhline(0, color='black', linestyle='--', alpha=0.5)
plt.xlabel('Log Age (Local) [yr]', fontsize=12)
plt.ylabel('Hubble Residual (mag)', fontsize=12)
plt.title('Figure 1: Age-Bias Correlation (Pantheon+ Basis)', fontsize=14, pad=15)
plt.grid(True, linestyle=':', alpha=0.5)

plt.text(0.05, 0.05, f"Pearson r = {r_value:.3f}\np-value = {p_value:.2e}", 
         transform=plt.gca().transAxes, fontsize=12, fontweight='bold', bbox=dict(facecolor='white', alpha=0.8))

plt.legend(loc='upper right', fontsize=10)
plt.tight_layout()
plt.savefig('fig1_replication_premium.png')
plt.show()
```
