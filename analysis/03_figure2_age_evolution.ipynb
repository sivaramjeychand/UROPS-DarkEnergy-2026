# Figure 2 Replication: Age Evolution

This notebook replicates Figure 2, showing the distribution of host galaxy ages as a function of redshift for the Rose et al. (2019) sample.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import os
import seaborn as sns

# Load Corrected Data
DATA_FILE = 'Rose19_corrected.csv'
df = pd.read_csv(DATA_FILE)

print(f"Loaded {len(df)} SNe.")
print(df[['SNID', 'z', 'logAl']].head())
```

## Plot Age vs Redshift
We plot the local age (`logAl`) against redshift (`z`).

```python
plt.figure(figsize=(10, 6))
plt.scatter(df['z'], df['logAl'], alpha=0.7, c=df['HR'], cmap='coolwarm', edgecolors='k')
plt.colorbar(label='Hubble Residual (mag)')
plt.xlabel('Redshift (z)')
plt.ylabel('Log Age (Local) [yr]')
plt.title('Host Galaxy Age vs Redshift (Rose+19)')
plt.grid(True, linestyle='--', alpha=0.5)

# Add trend line
z_range = np.linspace(df['z'].min(), df['z'].max(), 100)
p = np.polyfit(df['z'], df['logAl'], 1)
plt.plot(z_range, np.polyval(p, z_range), 'k--', label=f'Trend: slope={p[0]:.2f}')
plt.legend()

plt.savefig('fig2_replication.png')
plt.show()
```

## Histogram of Ages
Distribution of ages in the sample.

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['logAl'], bins=15, kde=True)
plt.xlabel('Log Age (Local) [yr]')
plt.title('Distribution of Host Galaxy Ages')
plt.show()
```
